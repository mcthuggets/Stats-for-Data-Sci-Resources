```python
import numpy as np
 
required = np.array([5, 4, 5])
 
employee_A = np.array([5, 3, 4])
employee_B = np.array([4, 5, 5])
 
score_A = np.dot(employee_A, required)
score_B = np.dot(employee_B, required)
 
print("Employee A Score:", score_A)
print("Employee B Score:", score_B)
 
if score_A > score_B:
    print("Employee A is the better match.")
else:
    print("Employee B is the better match.")
```
