```python
import matplotlib.pyplot as plt

data = [10, 12, 15, 15, 18, 20, 22, 25]

# Histogram
plt.hist(data, bins=5, edgecolor="black")
plt.title("Histogram")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()

# Boxplot
plt.boxplot(data)
plt.title("Boxplot")
plt.ylabel("Values")
plt.show()
```
