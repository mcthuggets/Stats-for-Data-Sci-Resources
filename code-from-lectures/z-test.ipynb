```python
from scipy.stats import norm
 
sample_mean = 52
population_mean = 50
population_sd = 4
sample_size = 10
 
z_stat = (sample_mean - population_mean) / (population_sd / (sample_size ** 0.5))
 
p_value = 2 * (1 - norm.cdf(abs(z_stat)))
 
print("Z-statistic =", z_stat)
print("P-value =", p_value)
 
if p_value < 0.05:
    print("Reject H0")
    print("Conclusion: The average weight is significantly different from 50 grams.")
else:
    print("Fail to reject H0")
    print("Conclusion: There is no significant difference from 50 grams.")
```
