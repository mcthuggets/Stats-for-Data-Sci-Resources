```python
from scipy.stats import ttest_1samp

scores = [52, 48, 51, 49, 53, 50, 54, 47, 51, 50]

t_stat, p_value = ttest_1samp(scores, 50)

print("T-statistic =", t_stat)
print("P-value =", p_value)

if p_value < 0.05:
    print("Reject H0")
    print("Conclusion: The average score is significantly different from 50.")
else:
    print("Fail to reject H0")
    print("Conclusion: There is no significant difference from 50.")
```
