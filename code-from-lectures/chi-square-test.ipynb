```python
from scipy.stats import chisquare
 
observed = [30, 20, 10]
expected = [20, 20, 20]
 
chi_stat, p_value = chisquare(observed, expected)
 
print("Chi-square statistic =",chi_stat)
print("P-value =", p_value)
 
if p_value < 0.05:
    print("Reject H0")
    print("Conclusion: Customer preferences are significantly different.")
else:
    print("Fail to reject H0")
    print("Conclusion: There is no significant difference in customer preferences.")
```
